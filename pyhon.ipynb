```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Eiabe\Desktop\vestuario\vendas_loja_nivel1.csv")
```

```python
df['cliente'] = df['cliente'].fillna("Desconhecido")
df = df.dropna(subset=['quantidade'])
```

```python
df['quantidade'] = df['quantidade'].astype(int)
df['data_venda'] = pd.to_datetime(df['data_venda'])
```

```python
duplicadas = ['produto', 'categoria', 'cliente', 'quantidade', 'preco_unitario', 'data_venda']
df = df[duplicadas].drop_duplicates()
```

```python
padronizacao = ['produto', 'categoria', 'cliente']

for i in padronizacao:
    df[i] = df[i].str.strip()
    df[i] = df[i].str.title()
```

```python
df['valor_venda'] = df['quantidade'] * df['preco_unitario']

faturamento_total = (df['quantidade'] * df['preco_unitario']).sum()
ranking_quantidade = df.groupby('produto')['quantidade'].sum().reset_index()
ranking_categoria = df.groupby('categoria')['valor_venda'].sum().reset_index()
```

```python
ranking_quantidade = ranking_quantidade.rename(columns={'quantidade': 'total_vendido'})
```

```python
ranking_quantidade['total_vendido'] = ranking_quantidade['total_vendido'].astype(int)

ranking_quantidade = ranking_quantidade.sort_values(by='total_vendido', ascending = False)

ranking_quantidade
```

```python
ranking_categoria['ranking'] = ranking_categoria['valor_venda'].rank(ascending=False, method= 'min')
ranking_categoria['ranking'] = ranking_categoria['ranking'].astype(int)
ranking_categoria = ranking_categoria.sort_values(by='ranking')


ranking_categoria
```
